Resolver Columnas Numéricas (8 por Mes)
</span>

```python
# Identify monthly consumption columns
pattern = re.compile(
    r"^(?P<month>[a-z]{3}\d{2})(?P<suffix>vta|sis|int|def|otro|tot|don|venc)$",
    re.IGNORECASE
)

monthly_columns = []

for column in df.columns:
    match = pattern.fullmatch(str(column))
    if match:
        monthly_columns.append({
            "Column": column,
            "Month": match.group("month"),
            "Suffix": match.group("suffix").lower()
        })

monthly_inventory = pd.DataFrame(monthly_columns)

expected_suffixes = [
    "vta", "sis", "int", "def", "otro", "tot", "don", "venc"
]

print("=" * 80)
print("MONTHLY COLUMN INVENTORY")
print("=" * 80)
print(f"Monthly columns found: {len(monthly_inventory)}")
print(f"Expected monthly columns: {12 * 8}")
print(f"Distinct months found: {monthly_inventory['Month'].nunique()}")
print("\nColumns by month:")

if not monthly_inventory.empty:
    month_summary = (
        monthly_inventory.groupby("Month")["Suffix"]
        .agg(
            Column_count="count",
            Suffixes=lambda values: ", ".join(sorted(values))
        )
        .reset_index()
    )
    print(month_summary.to_string(index=False))

    print("\nMissing expected columns:")
    month_suffix_pairs = set(
        zip(monthly_inventory["Month"], monthly_inventory["Suffix"])
    )
    months = sorted(monthly_inventory["Month"].unique())

    missing_columns = [
        f"{month}{suffix}"
        for month in months
        for suffix in expected_suffixes
        if (month, suffix) not in month_suffix_pairs
    ]

    if missing_columns:
        print("\n".join(missing_columns))
    else:
        print("No missing suffixes detected for the months found.")
else:
    print("No columns matched the expected naming pattern.")
    print("Review the actual column names before proceeding.")

print("=" * 80)
```

```python
suffixes = ["vta", "sis", "int", "def", "otro", "tot", "don", "venc"]

column_pattern = re.compile(
    r"^([a-z]{3})(\d{2})(vta|sis|int|def|otro|tot|don|venc)$",
    re.IGNORECASE
)

monthly_columns = {}

for column in df.columns:
    match = column_pattern.fullmatch(str(column))

    if match:
        month_code = match.group(1).lower()
        year_code = int(match.group(2))
        suffix = match.group(3).lower()

        month_key = f"{month_code}{year_code:02d}"
        monthly_columns.setdefault(month_key, {})[suffix] = column

# Sort from newest to oldest, assuming two-digit years represent 2000-2099
months = sorted(
    monthly_columns,
    key=lambda month: (
        int(month[-2:]),
        ["ene", "feb", "mar", "abr", "may", "jun",
         "jul", "ago", "set", "oct", "nov", "dic"].index(month[:3])
        if month[:3] in [
            "ene", "feb", "mar", "abr", "may", "jun",
            "jul", "ago", "set", "oct", "nov", "dic"
        ] else -1
    ),
    reverse=True
)

print("=" * 80)
print("MONTHLY COLUMN INVENTORY")
print("=" * 80)
print(f"Months detected: {len(months)}")
print(f"Monthly columns detected: {sum(len(monthly_columns[m]) for m in months)}")
print(f"Expected columns: {12 * 8}")
print("\nMonths and available suffixes:")

for month in months:
    available = monthly_columns[month]
    missing = [suffix for suffix in suffixes if suffix not in available]

    print(
        f"{month}: {len(available)}/8 columns"
        + (f" | Missing: {', '.join(missing)}" if missing else "")
    )

print("=" * 80)
```

```python
components = ["vta", "sis", "int", "def", "otro"]
validation_results = []

for month in months:
    columns = monthly_columns[month]

    required = components + ["tot"]
    missing_columns = [suffix for suffix in required if suffix not in columns]

    if missing_columns:
        validation_results.append({
            "Month": month,
            "Rows_checked": 0,
            "Total_mismatches": 0,
            "Incomplete_rows": len(df),
            "Invalid_numeric_values": 0,
            "Infinite_values": 0,
            "Negative_values": 0,
            "Status": "MISSING REQUIRED COLUMNS"
        })
        continue

    numeric_data = pd.DataFrame(index=df.index)
    invalid_numeric = pd.Series(False, index=df.index)
    infinite_values = pd.Series(False, index=df.index)
    negative_values = pd.Series(False, index=df.index)

    for suffix in suffixes:
        column = columns.get(suffix)

        if column is None:
            continue

        original = df[column]
        converted = pd.to_numeric(original, errors="coerce")

        invalid_numeric |= original.notna() & converted.isna()

        numeric_data[suffix] = converted

        values = converted.to_numpy(dtype=float, na_value=np.nan)
        infinite_values |= pd.Series(np.isinf(values), index=df.index)
        negative_values |= converted.lt(0).fillna(False)

    # Only finite values are used in arithmetic checks
    for suffix in components + ["tot"]:
        numeric_data[suffix] = numeric_data[suffix].where(
            np.isfinite(numeric_data[suffix].to_numpy(dtype=float, na_value=np.nan))
        )

    complete_components = numeric_data[components].notna().all(axis=1)
    valid_total = numeric_data["tot"].notna()

    calculated_total = numeric_data[components].sum(
        axis=1,
        min_count=len(components)
    )

    comparable = complete_components & valid_total

    difference = numeric_data["tot"] - calculated_total
    mismatch = comparable & ~np.isclose(
        numeric_data["tot"].fillna(0),
        calculated_total.fillna(0),
        rtol=1e-9,
        atol=1e-6
    )

    incomplete = ~comparable

    validation_results.append({
        "Month": month,
        "Rows_checked": int(comparable.sum()),
        "Total_mismatches": int(mismatch.sum()),
        "Incomplete_rows": int(incomplete.sum()),
        "Invalid_numeric_values": int(invalid_numeric.sum()),
        "Infinite_values": int(infinite_values.sum()),
        "Negative_values": int(negative_values.sum()),
        "Status": "REVIEW REQUIRED" if (
            mismatch.any()
            or incomplete.any()
            or invalid_numeric.any()
            or infinite_values.any()
            or negative_values.any()
        ) else "OK"
    })

    # Save detailed discrepancies for this month
    if mismatch.any():
        detail = pd.DataFrame(index=df.index[mismatch])
        detail["Month"] = month

        for suffix in required:
            detail[columns[suffix]] = df.loc[mismatch, columns[suffix]]

        detail["Calculated_total"] = calculated_total.loc[mismatch]
        detail["Difference"] = difference.loc[mismatch]
        detail["Original_index"] = df.index[mismatch]

        if "monthly_mismatches" not in locals():
            monthly_mismatches = []

        monthly_mismatches.append(detail.reset_index(drop=True))

summary = pd.DataFrame(validation_results)

print("=" * 100)
print("MONTHLY TOTAL VALIDATION")
print("=" * 100)
print(summary.to_string(index=False))
print("=" * 100)
```

```python
if "monthly_mismatches" in locals() and monthly_mismatches:
    mismatch_report = pd.concat(monthly_mismatches, ignore_index=True)

    print(f"Total discrepancies found: {len(mismatch_report)}")
    print(mismatch_report.to_string(index=False))
else:
    mismatch_report = pd.DataFrame()
    print("No total discrepancies were found among comparable rows.")
```

<span style="font-family: 'Glass Antiqua', serif; font-size: 26px;">
Asignar Abreviaturas y Nombres
</span>

```python
# 1. Preserve original values
if "subtipo_original" not in df.columns:
    df["subtipo_original"] = df["subtipo"].copy()

if "Nomsubtipo_original" not in df.columns:
    df["Nomsubtipo_original"] = df["Nomsubtipo"].copy()

# 2. Normalize subtype codes
subtype_normalization = {
    "M": "M",
    "MEDICAMENTO": "M",
    "MEDICAMENTOS": "M",
    "INSUMOS": "I",
    "I": "I",
    "R": "R",
    "POR DEFINIR": "SD",
    "SD": "SD",
    "PB": "PB",
    "IM": "IM",
    "PG": "PG",
    "PS": "PS",
    "PD": "PD",
    "AYDE": "AD",
    "AD": "AD",
    "O": "OT",
    "OT": "OT",
    "IO": "IO",
    "NP": "NP",
    "A": "AL",
    "AL": "AL",
    "ADX": "ADX",
    "MQ": "MQ",
    "L": "SLM",
    "SLM": "SLM",
    "NE": "NE",
    "PN": "PN",
    "PR": "PR",
}

df["subtipo"] = (
    df["subtipo"]
    .astype("string")
    .str.strip()
    .str.upper()
    .replace(subtype_normalization)
    .fillna("SD")
)

# 3. Define standardized subtype descriptions
subtype_names = {
    "M": "MEDICAMENTO",
    "I": "INSUMOS",
    "R": "REACTIVOS Y MATERIAL DE LABORATORIO",
    "SD": "SIN DEFINIR",
    "PB": "PRODUCTO BIOLOGICO",
    "IM": "INSTRUMENTAL MEDICO",
    "PG": "PRODUCTO GALENICO",
    "PS": "PRODUCTO SANITARIO",
    "PD": "PRODUCTO DIETETICO",
    "AD": "ANTISEPTICO Y DESINFECTANTE",
    "OT": "OTROS",
    "IO": "INSTRUMENTAL ODONTOLOGICO",
    "NP": "NUTRICION PARENTERAL",
    "AL": "ALIMENTOS, CEREALES, FORMULAS Y NUTRIENTES",
    "ADX": "AGENTE DE DIAGNOSTICO",
    "MQ": "MEDICAMENTOS QUIMICOS",
    "SLM": "SUCEDANEOS DE LA LECHE MATERNA",
    "NE": "NUTRICION ENTERAL",
    "PN": "PRODUCTO NATURAL",
    "PR": "PRODUCTO RADIOFARMACO",
}

# 4. Assign standardized descriptions based on subtype codes
df["Nomsubtipo"] = df["subtipo"].map(subtype_names).fillna(
    "REVISAR CLASIFICACION"
)

# 5. Generate the final validation summary
validation = (
    df.groupby(["subtipo", "Nomsubtipo"], dropna=False)
    .size()
    .reset_index(name="Count")
    .sort_values("subtipo")
)

print("=" * 90)
print("FINAL SUBTYPE CATALOG")
print("=" * 90)
print(validation.to_string(index=False))

print("\n" + "=" * 90)
print("VALIDATION SUMMARY")
print("=" * 90)
print(f"Total records: {len(df):,}")
print(f"Distinct subtype codes: {df['subtipo'].nunique()}")
print(f"Undefined records (SD): {(df['subtipo'] == 'SD').sum():,}")
print(
    "Records requiring classification review: "
    f"{(df['Nomsubtipo'] == 'REVISAR CLASIFICACION').sum():,}"
)
print("=" * 90)
```
