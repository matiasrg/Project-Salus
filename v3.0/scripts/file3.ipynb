Resolver Columnas Numéricas (8 por Mes)
</span>

```python
# Detect monthly columns and sort periods chronologically.
suffixes = ["vta", "sis", "int", "def", "otro", "tot", "don", "venc"]

months = [
    "ene", "feb", "mar", "abr", "may", "jun",
    "jul", "ago", "set", "oct", "nov", "dic",
]

month_order = {
    month: position
    for position, month in enumerate(months)
}

month_pattern = "|".join(months)
suffix_pattern = "|".join(suffixes)

pattern = re.compile(
    rf"^(?P<month>{month_pattern})(?P<year>\d{{2}})"
    rf"(?P<suffix>{suffix_pattern})$",
    re.IGNORECASE,
)
```

```python
monthly_columns = {}

for column in df.columns:
    match = pattern.fullmatch(str(column))

    if match:
        period = f"{match['month'].lower()}{match['year']}"
        suffix = match["suffix"].lower()
        monthly_columns.setdefault(period, {})[suffix] = column

months = sorted(
    monthly_columns,
    key=lambda period: (
        int(period[-2:]),
        month_order.get(period[:3], -1),
    ),
    reverse=True,
)
```

```python
# Review monthly columns in chronological order.
inventory = pd.DataFrame([
    {
        "Month": month,
        "Columns_found": len(columns),
        "Missing_suffixes": ", ".join(
            s for s in suffixes if s not in columns
        ) or "None",
    }
    for month, columns in sorted(
        monthly_columns.items(),
        key=lambda item: (
            int(item[0][-2:]),
            month_order.get(item[0][:3], -1),
        ),
        reverse=True,
    )
])

line = "=" * 65

print(f"\n{line}")
print("MONTHLY COLUMN VALIDATION")
print(line)
print(f"Months detected        : {len(monthly_columns)}")
print(f"Monthly columns found  : {sum(map(len, monthly_columns.values()))}")
print(f"Expected for 12 months : {12 * len(suffixes)}")
print("-" * 65)
print(inventory.to_string(index=False))
print(line)
```

```python
# Validate monthly totals and flag invalid values.
components = ["vta", "sis", "int", "def"]
required = components + ["tot"]
validation_results = []
monthly_mismatches = []

for month in months:
    columns = monthly_columns[month]
    missing = [s for s in required if s not in columns]

    if missing:
        validation_results.append({
            "Month": month,
            "Status": f"MISSING COLUMNS: {', '.join(missing)}",
        })
        continue

    values = {}
    invalid = infinite = negative = pd.Series(False, index=df.index)

    for suffix in required:
        original = df[columns[suffix]]
        numeric = pd.to_numeric(original, errors="coerce")
        array = numeric.to_numpy(dtype=float, na_value=np.nan)

        invalid |= original.notna() & numeric.isna()
        infinite |= pd.Series(np.isinf(array), index=df.index)
        negative |= numeric.lt(0).fillna(False)
        values[suffix] = numeric.where(np.isfinite(array))

    numeric_df = pd.DataFrame(values)
    complete = numeric_df.notna().all(axis=1)
    calculated = numeric_df[components].sum(
        axis=1, min_count=len(components)
    )
    difference = values["tot"] - calculated

    mismatch = complete & ~np.isclose(
        values["tot"].fillna(0),
        calculated.fillna(0),
        rtol=1e-9,
        atol=1e-6,
    )

    issues = (
        mismatch.any() or (~complete).any() or invalid.any()
        or infinite.any() or negative.any()
    )

    validation_results.append({
        "Month": month,
        "Rows_checked": int(complete.sum()),
        "Total_mismatches": int(mismatch.sum()),
        "Incomplete_rows": int((~complete).sum()),
        "Invalid_numeric_values": int(invalid.sum()),
        "Infinite_values": int(infinite.sum()),
        "Negative_values": int(negative.sum()),
        "Status": "REVIEW REQUIRED" if issues else "OK",
    })

    if mismatch.any():
        detail = df.loc[mismatch, [columns[s] for s in required]].copy()
        detail.insert(0, "Month", month)
        detail["Calculated_total"] = calculated[mismatch]
        detail["Difference"] = difference[mismatch]
        monthly_mismatches.append(detail.reset_index(drop=True))

# Print validation summary.
summary = pd.DataFrame(validation_results)
line = "=" * 128

print(f"\n{line}")
print("MONTHLY TOTALS VALIDATION")
print(line)
print(f"Months validated : {len(validation_results)}")
print(f"Months requiring review : {(summary['Status'] != 'OK').sum()}")
print("-" * 128)
print(summary.to_string(index=False))
print(line)
```

```python
# Report monthly total discrepancies.
if monthly_mismatches:
    mismatch_report = pd.concat(monthly_mismatches, ignore_index=True)
    print(f"Total discrepancies found: {len(mismatch_report)}")
    print(mismatch_report.to_string(index=False))
else:
    print("No total discrepancies found.")
```

```python
# UPDATE MONTHLY TOTALS

for month in months:
    columns = monthly_columns[month]
    component_columns = [columns[suffix] for suffix in components]

    df[columns["tot"]] = (
        df[component_columns]
        .apply(pd.to_numeric, errors="coerce")
        .sum(axis=1, min_count=len(components))
    )

print("Monthly totals updated.")
```

<span style="font-family: 'Glass Antiqua', serif; font-size: 26px;">
Asignar Abreviaturas y Nombres
</span>

```python
subtype_normalization = {
    "M": "M",
    "MEDICAMENTO": "M",
    "MEDICAMENTOS": "M",
    "INSUMOS": "I",
    "I": "I",
    "R": "R",
    "POR DEFINIR": "SD",
    "SD": "SD",
    "PB": "PB",
    "IM": "IM",
    "PG": "PG",
    "PS": "PS",
    "PD": "PD",
    "AYDE": "AD",
    "AD": "AD",
    "O": "OT",
    "OT": "OT",
    "IO": "IO",
    "NP": "NP",
    "A": "AL",
    "AL": "AL",
    "ADX": "ADX",
    "MQ": "MQ",
    "L": "SLM",
    "SLM": "SLM",
    "NE": "NE",
    "PN": "PN",
    "PR": "PR",
}

df["subtipo"] = (
    df["subtipo"]
    .astype("string")
    .str.strip()
    .str.upper()
    .replace(subtype_normalization)
    .fillna("SD")
)

subtype_names = {
    "M": "MEDICAMENTO",
    "I": "INSUMOS",
    "R": "REACTIVOS Y MATERIAL DE LABORATORIO",
    "SD": "SIN DEFINIR",
    "PB": "PRODUCTO BIOLOGICO",
    "IM": "INSTRUMENTAL MEDICO",
    "PG": "PRODUCTO GALENICO",
    "PS": "PRODUCTO SANITARIO",
    "PD": "PRODUCTO DIETETICO",
    "AD": "ANTISEPTICO Y DESINFECTANTE",
    "OT": "OTROS",
    "IO": "INSTRUMENTAL ODONTOLOGICO",
    "NP": "NUTRICION PARENTERAL",
    "AL": "ALIMENTOS, CEREALES, FORMULAS Y NUTRIENTES",
    "ADX": "AGENTE DE DIAGNOSTICO",
    "MQ": "MEDICAMENTOS QUIMICOS",
    "SLM": "SUCEDANEOS DE LA LECHE MATERNA",
    "NE": "NUTRICION ENTERAL",
    "PN": "PRODUCTO NATURAL",
    "PR": "PRODUCTO RADIOFARMACO",
}

df["Nomsubtipo"] = df["subtipo"].map(subtype_names).fillna(
    "REVISAR CLASIFICACION"
)

validation = (
    df.groupby(["subtipo", "Nomsubtipo"], dropna=False)
    .size()
    .reset_index(name="Count")
    .sort_values("subtipo")
)

print("=" * 90)
print("FINAL SUBTYPE CATALOG")
print("=" * 90)
print(validation.to_string(index=False))

print("\n" + "=" * 90)
print("VALIDATION SUMMARY")
print("=" * 90)
print(f"Total records: {len(df):,}")
print(f"Distinct subtype codes: {df['subtipo'].nunique()}")
print(f"Undefined records (SD): {(df['subtipo'] == 'SD').sum():,}")
print(
    "Records requiring classification review: "
    f"{(df['Nomsubtipo'] == 'REVISAR CLASIFICACION').sum():,}"
)
print("=" * 90)
```
